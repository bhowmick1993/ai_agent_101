```python
from langchain.agents import create_agent
from langchain_huggingface import HuggingFacePipeline, ChatHuggingFace

```

```python
# 1. Load the local Hugging Face model
llm = HuggingFacePipeline.from_model_id(
    model_id="Qwen/Qwen3-4B",
    task="text-generation",

    model_kwargs={
        "torch_dtype": "auto",
        "device_map": "auto",
    },

    pipeline_kwargs={
        "max_new_tokens": 512,
        "do_sample": False,
        "return_full_text": False,
    },
)
```

```python
# 2. Convert it into a LangChain chat model
model = ChatHuggingFace(llm=llm)
```

```python
# 3. Create the LangChain agent
agent = create_agent(
    model=model,
    tools=[],
    system_prompt=(
        "You are a helpful assistant. "
        "Solve problems carefully and explain the answer clearly."
    ),
)
```

```python
# 4. Run the agent
result = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": """
Calculate how far a vehicle travels in 5 seconds
if it is moving at 72 km/h.
Explain your reasoning.
""",
            }
        ]
    }
)
```
